```python
import pandas as pd

CAMINHO_CSV = "Plant_2_Generation_Data.csv"  # ajustar para o nome/caminho real do arquivo exportado do Orange

df = pd.read_csv(CAMINHO_CSV)
df.columns
```

### 1. Renomear variáveis principais

```python
df = df.rename(columns={
    "DC_POWER": "Potencia_CC",
    "AC_POWER": "Potencia_CA",
    "DAILY_YIELD": "Geracao_Diaria",
})
df.head()
```

### 3. Maior potência CA registrada

```python
max_potencia_ca = df["Potencia_CA"].max()
max_potencia_ca
```

### 4. Limiar de 70% da potência CA máxima

```python
limiar_70 = 0.70 * max_potencia_ca
limiar_70
```

### 5. DataFrame de alta geração: quantidade e percentual

```python
df_alta_geracao = df[df["Potencia_CA"] > limiar_70]

qtd_alta_geracao = len(df_alta_geracao)
percentual_alta_geracao = qtd_alta_geracao / len(df) * 100

print(f"Registros de alta geração: {qtd_alta_geracao}")
print(f"Percentual sobre o total da amostra: {percentual_alta_geracao:.2f}%")
```

### 6. value_counts() em SOURCE_KEY dentro da alta geração

```python
contagem_inversores = df_alta_geracao["SOURCE_KEY"].value_counts()
contagem_inversores
```

### 7. Inversor mais frequente e interpretação

```python
inversor_mais_frequente = contagem_inversores.idxmax()
qtd_mais_frequente = contagem_inversores.max()
percentual_do_inversor = qtd_mais_frequente / qtd_alta_geracao * 100

print(f"Inversor mais frequente em alta geração: {inversor_mais_frequente}")
print(f"Aparece em {qtd_mais_frequente} dos {qtd_alta_geracao} registros de alta geração ({percentual_do_inversor:.2f}%)")
```

**Interpretação (preencher com os valores impressos acima antes de entregar):**

O inversor `<preencher com o valor de inversor_mais_frequente>` é o `SOURCE_KEY` que mais aparece nos registros acima de 70% da potência CA máxima, respondendo por `<preencher percentual>`% desses registros.

Isso indica apenas **frequência de ocorrência acima do limiar** — ou seja, esse inversor está associado a um número maior de intervalos de tempo em que a potência CA da planta como um todo ultrapassou 70% do seu máximo. Não é possível, só com esse recorte, concluir que esse inversor específico é "melhor" ou "mais eficiente":

- A amostra é de 20% dos registros da planta inteira, não de um inversor isolado — o `SOURCE_KEY` mais frequente pode simplesmente ter mais leituras registradas no período coberto pela amostra (viés de contagem, não de desempenho).
- Não há normalização por tempo de operação, irradiância solar no momento da leitura, nem comparação com o próprio máximo de cada inversor individualmente — o limiar de 70% foi calculado sobre a potência CA da coluna agregada, não por inversor.
- Um inversor pode aparecer mais vezes acima do limiar só porque está fisicamente em uma posição com menos sombreamento ou horário de leitura diferente, não porque tem melhor performance intrínseca.

Para afirmar desempenho ou falha, seria necessário comparar cada inversor com o seu próprio histórico/capacidade nominal, controlar por horário e irradiância, e olhar a série temporal completa — não apenas a contagem de aparições acima de um limiar global.
