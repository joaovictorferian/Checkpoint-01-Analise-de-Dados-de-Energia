### 2. Renomear Appliances e simplificar nomes ambientais

Ajuste os nomes originais de temperatura/umidade (`T1`, `T2`, `RH_1`, ...) conforme as colunas mantidas na Etapa A — o exemplo abaixo assume `T1`, `T2`, `T3` e `RH_1`, `RH_2`, `RH_3`.

```python
df = df.rename(columns={
    "Appliances": "Consumo_Eletrodomesticos",
    "T1": "Temp_Sala",
    "T2": "Temp_Escritorio",
    "T3": "Temp_Cozinha",
    "RH_1": "Umid_Sala",
    "RH_2": "Umid_Escritorio",
    "RH_3": "Umid_Cozinha",
})
df.head()
```

### 3. Maior consumo de eletrodomésticos registrado

```python
max_consumo = df["Consumo_Eletrodomesticos"].max()
max_consumo
```

### 4. Limiar de 70% do máximo e DataFrame de consumo elevado

```python
limiar_70 = 0.70 * max_consumo
df_consumo_alto = df[df["Consumo_Eletrodomesticos"] > limiar_70]
df_consumo_alto.head()
```

### 5. Quantidade e percentual de registros selecionados

```python
qtd_consumo_alto = len(df_consumo_alto)
percentual_consumo_alto = qtd_consumo_alto / len(df) * 100

print(f"Registros de consumo elevado: {qtd_consumo_alto}")
print(f"Percentual sobre o total da amostra: {percentual_consumo_alto:.2f}%")
```

### 6. Temperatura média de T1 e segundo DataFrame (consumo alto E temperatura acima da média)

```python
temp_media = df["Temp_Sala"].mean()
temp_media
```

```python
df_consumo_temp_alta = df[
    (df["Consumo_Eletrodomesticos"] > limiar_70) &
    (df["Temp_Sala"] > temp_media)
]

qtd_consumo_temp_alta = len(df_consumo_temp_alta)
percentual_consumo_temp_alta = qtd_consumo_temp_alta / len(df) * 100

print(f"Registros com consumo alto E temperatura acima da média: {qtd_consumo_temp_alta}")
print(f"Percentual sobre o total da amostra: {percentual_consumo_temp_alta:.2f}%")
```

### 7. Comparação entre os dois DataFrames

```python
print(f"DataFrame 1 (só consumo elevado): {qtd_consumo_alto} registros")
print(f"DataFrame 2 (consumo elevado E temperatura acima da média): {qtd_consumo_temp_alta} registros")
print(f"Redução ao adicionar o critério de temperatura: {qtd_consumo_alto - qtd_consumo_temp_alta} registros a menos")
```

**Interpretação (preencher com os valores impressos acima antes de entregar):**

O DataFrame 2 tem `<preencher>` registros, contra `<preencher>` do DataFrame 1 — ou seja, adicionar a condição de temperatura acima da média **reduz** o conjunto (interseção de condições nunca aumenta o total; no máximo, mantém).

Isso mostra que nem todo período de consumo elevado de eletrodomésticos coincide com temperatura ambiente acima da média — parte dos picos de consumo ocorre em momentos de temperatura mais baixa (possivelmente ligados a aquecedores, ou uso de eletrodomésticos sem relação direta com a temperatura da sala, como máquina de lavar). O segundo critério funciona como um filtro mais restritivo, útil para investigar especificamente os casos em que consumo elevado e ambiente mais quente ocorrem juntos, mas não descreve a totalidade dos picos de consumo da residência.
